crosstab:A crosstabulation (cross-tab) is a Pandas method used to create a frequency table that summarizes the relationship between two or more categorical variables.

```python
import pandas as pd
data=pd.read_csv("/content/sample_data/cars.csv")
data
```

```python
pd.crosstab(data[' cubicinches'],data[' cylinders'])
```

```python
pd.crosstab(data[' hp'],data[' year'],margins=True)
```

```python
pd.crosstab(data[' brand'],data[' weightlbs'])
```

```python
pd.crosstab(data[' brand'],data[" cylinders"],values=data[" cubicinches"],aggfunc='min')
```

```python
pd.crosstab(data[' brand'], data[' cylinders'],normalize=True,margins=True)*100
```

```python
table=pd.crosstab(data[' brand'], data[' cylinders'],normalize='columns')*100
```

```python
table
```

```python
pd.crosstab(data[' brand'], data[' cylinders'],normalize='index')*100
```

```python
pd.crosstab(data[' brand'],data[' cylinders'],values=data[" hp"],aggfunc='count',margins=True)
```

```python
table=pd.crosstab(data[' brand'],data[' cylinders'])
table.plot(kind='barh',stacked=True)
```
