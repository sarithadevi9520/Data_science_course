Handling Null Values

```python
data = {
    'Name': ['Ravi', 'Priya', 'Arun', 'Sneha', 'Kiran', 'Anu'],
    'Age': [25, np.nan, 24, 30, np.nan, 29],
    'Salary': [30000, 45000, np.nan, 55000, 40000, np.nan],
    'Department': ['IT', np.nan, 'HR', np.nan, 'HR', 'IT']
}
df = pd.DataFrame(data)
df
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
df.isnull().any()
```

```python
df['Age'].isnull()
```

```python
df[df['Age'].isnull()]
```

```python
df['Age'].fillna(28)
```

```python
df['Department'].fillna('Unknown')
```

```python
df['Salary']
```

```python
avg_sal=df['Salary'].mean()
avg_sal
median_age=df['Age'].median()
median_age
mode_dept=df['Department'].mode()[1]
mode_dept
```

```python
df.fillna({'Salary':avg_sal,'Age':median_age,'Department':mode_dept},inplace=True)
```
