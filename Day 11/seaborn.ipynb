```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
```

```python
tips
```

```python
sns.get_dataset_names()
```

```python
tips.head()
```

```python
pd.crosstab(tips['day'],tips['time']).plot(kind='bar')
```

```python
tip=pd.crosstab(tips['sex'],tips['day'])
tip.plot(kind='bar')
```

```python
sns.stripplot(y='tip',data=tips)
```

```python
sns.stripplot(y='tip',data=tips,jitter=True)
```

```python
sns.stripplot(x='sex',y='tip',data=tips,hue='sex')
```

```python
sns.stripplot(x='total_bill',y='sex',data=tips,jitter=True)
```

```python
sns.stripplot(x='tip',y='day',data=tips,size=6,jitter=True,hue='day')
```

```python
sns.swarmplot(data=tips,x='day',y='tip',hue='day',palette='Set2')
```

```python
sns.swarmplot(x='day',y='tip',hue='sex',data=tips)
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
sns.boxplot(data=tips,x='day',y='tip',hue='sex',palette='Set2')

plt.subplot(2,2,2)
sns.boxplot(data=tips,x='sex',y='total_bill')

plt.subplot(2,2,3)
sns.violinplot(data=tips,x='smoker',y='total_bill')

plt.subplot(2,2,4)
sns.boxplot(data=tips,x='sex',y='tip')
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
sns.histplot(data=tips,x='total_bill')
```

```python
sns.jointplot(data=tips,x='total_bill',y='tip',kind='kde',hue='sex')
```

```python
sns.jointplot(data=tips,x='tip',y='day',kind='scatter')
```

```python


pd.plotting.scatter_matrix(tips)
plt.show()
```

```python
sns.pairplot(tips)
```

```python
sns.pairplot(tips,hue='day')
```
