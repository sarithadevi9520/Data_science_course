Matplotlib Analysis

```python
plot_data = high_sales.reset_index()
plot_data
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=plot_data,
    x='sub_category',
    y='total_price',
    hue='category'
)

plt.title('Total Sales by Category and Sub-category')
plt.xlabel('Sub-category')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('category')['total_price'].sum()

plt.figure(figsize=(8,5))

plt.bar(category_sales.index, category_sales.values)

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')

plt.show()
```

```python
region_sales = df.groupby('region')['total_price'].sum()

plt.figure(figsize=(8,5))

plt.bar(region_sales.index, region_sales.values)

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')

plt.show()
```

```python
monthly_sales = df.groupby(
    ['month', 'month_name']
)['total_price'].sum().sort_index()

plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales.index.get_level_values('month_name'),
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.show()
```

```python
day_sales = df.groupby(
    ['day_name', 'category']
)[['total_price', 'quantity']].sum()
```

```python
plot_day = day_sales.reset_index()

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

plot_day['day_name'] = pd.Categorical(
    plot_day['day_name'],
    categories=day_order,
    ordered=True
)

plot_day = plot_day.sort_values('day_name')

plt.figure(figsize=(12, 6))

sns.barplot(
    data=plot_day,
    x='day_name',
    y='total_price',
    hue='category'
)

plt.title('Total Sales by Day and Category')
plt.xlabel('Day')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```
