```python
import matplotlib.pyplot as plt
```

```python
x=[2023,2024,2025,2026,2027,2028]
y=[15,25,45,27,46,34]
y1=[23,5,6,78,34,45]
line_style=dict(marker="*",ms=10,markerfacecolor="red",markeredgecolor="green",linestyle="-.",linewidth=3)
plt.plot(x,y,color="blue",**line_style)
plt.plot(x,y1,color="green",**line_style)
plt.show()
```

```python
x=[2023,2024,2025,2026,2027,2028]
y=[15,25,45,27,46,34]
y1=[23,5,6,78,34,45]
y2=[34,55,67,34,34,56]

plt.title("Class Size",
          color="black",
          fontsize="20",
          fontfamily="Sans",
          fontweight="bold")
plt.xlabel("year",fontsize=10,fontfamily="DejaVu Sans",fontweight="bold")
plt.ylabel("Students",fontsize=10,fontfamily="DejaVu Sans",fontweight="bold")
plt.tick_params(axis="both",colors="blue")
plt.plot(x,y)
plt.plot(x,y1)
plt.plot(x,y2)
plt.show()
```

**Grid Lines**:grid() helps make plots easier to read by adding reference line

```python
x=[2023,2024,2025,2026,2027,2028]
y=[15,25,45,27,46,34]
plt.grid(linewidth=2,linestyle=":")
plt.plot(x,y)
plt.show()
```

**Bar chart** :compare categories of data by representing each category with a bar

```python
import numpy as np
dept=np.array(["hr","IT","Finance","Sales"])
salary=np.array([20000,35000,40000,25000])
plt.bar(dept,salary,color="skyblue",width=0.5)
#plt.barh(dept,salary,color="skyblue")
plt.title("Dept wise Salaries")
plt.xlabel("Dept")
plt.ylabel("Salary")
plt.show()
```

**Pie Chart**

```python
dept=np.array(["hr","IT","Finance","Sales"])
salary=np.array([20000,35000,40000,25000])
colors=["red","yellow","blue","pink"]
plt.pie(salary,
        labels=dept,
        autopct="%1.1f%%",
        colors=colors,
        explode=[0.3,0,0,0],
        shadow=True,
        startangle=90)
plt.show()
```

**Scatter Graphs**:shows the relationship between two variables
                   helps to identify a correlation(+tive,-tive,none)

```python
x1=[0,1,1,2,3,4,5,6,6,7,8,8,9]
y1=[55,60,63,68,69,70,73,78,79,80,82,85,90]

x2=[0,1,1,2,3,4,5,6,6,7,8,8,9]
y2=[50,59,64,70,75,76,79,80,89,90,92,55,99]

plt.scatter(x1,y1,
            color="green",
            alpha=0.9,
            s=50,
            label="class A")

plt.scatter(x2,y2,
            color="blue",
            alpha=0.9,
            s=50,
            label="class B")

plt.title("Test scores")
plt.xlabel("hours studied")
plt.ylabel("grade")
plt.legend()
plt.show()
```

**Histograms**:A visual representation of the distribution of quantitave data
             they group values into bins(intervals)

```python
scores=np.random.normal(loc=80,scale=10,size=100)
scores
```

```python
plt.hist(scores,
         bins=5,
         color="pink",
         edgecolor="black")
plt.title("exam scores")
plt.xlabel("Score")
plt.ylabel("no.of students")
plt.show()
```

**sub plots**:Figure=the entire figure,Axes=A single ploT(sub plots)

```python
x=np.array([1,2,3,4,5,6])
figure,axes=plt.subplots(2,2)
axes[0,0].plot(x,x*2,color="red")
axes[0,0].set_title("x*2")

axes[0,1].plot(x,x**2,color="green")
axes[0,1].set_title("x**2")

axes[1][0].scatter(x,x**3,color="blue")
axes[1,0].set_title("x*3")

axes[1][1].bar(x,x**4,color="blue")
axes[1,1].set_title("x**4")


plt.show()
```
