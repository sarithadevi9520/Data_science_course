Handling null values

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Name': ['A', np.nan, 'C'],
    'Age': [20, np.nan, 22],
    'Marks': [80, 90, np.nan]
})
data=pd.DataFrame(data)
data
```

```python
data.fillna(0)
```

```python
data.isnull()
```

```python
data.isnull().sum()
```

```python
mean1=data['Age'].mean()
int(mean1)
```

```python
median1=data['Marks'].median()
median1
```

```python
mode1=data['Name'].mode()
mode1
```

```python
mode1=data['Name'].mode()[0]
```

```python
dframe=data.fillna({'Age':mean1,'Marks':median1,'Name':mode1})
dframe
```

```python
data
```

Connecting Data frames

```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['A', 'B'],
    'Age': [20, 21]
})

df2 = pd.DataFrame({
    'Name': ['C', 'D'],
    'Age': [22, 23]
})
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index=True)
```

```python
customers = pd.DataFrame({
    'CustomerID': [1, 2, 3],
    'Name': ['Ravi', 'Sita', 'John']
})
```

```python
orders = pd.DataFrame({
    'CustomerID': [1, 2, 3],
    'Amount': [500, 700, 300]
})
```

```python
pd.merge(customers,orders,on="CustomerID")
```

```python
pd.merge(customers,orders,on="CustomerID",how="inner")
```

```python
pd.merge(customers,orders,on="CustomerID",how="outer")
```

```python
d1 = {'movieid':[1,2,3,4,5],
      'title':['Leo','Jawan','Gadar','Pathan','Uri']}
movies = pd.DataFrame(d1)
```

```python
d2 = {'memid':[1,2,3,4],
      'name':['Tom','Tim','Jon','Jim'],
      'movieid':[1,np.nan,3,5]}
members = pd.DataFrame(d2)
```

```python
pd.merge(movies,members,on="movieid")
```

```python
pd.merge(movies,members,on="movieid",how="inner")
```

```python
pd.merge(movies,members,on="movieid",how="outer")
```

```python
pd.merge(movies,members,on="movieid",how="left")
```

```python
pd.merge(movies,members,on="movieid",how="right")
```
