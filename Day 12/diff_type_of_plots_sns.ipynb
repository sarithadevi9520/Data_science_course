**Relational Plots in Seaborn**

Relational plots in Seaborn are used to visualize the relationship between two or more variables.

They help us answer questions like:

Does one variable increase when another increases?
Is there a relationship between two numerical variables?
How does a variable change across categories?

```python
import seaborn as sns
df=sns.load_dataset('tips')
df.head(3)
```

```python
sns.lineplot(data=df,x='size',y='tip')
```

```python
sns.scatterplot(data=df,x='total_bill',y='tip',hue='sex')
```

```python
sns.relplot(data=df,x='total_bill',y='tip',hue='sex')
```

**Categorical plots** in Seaborn are used to visualize the relationship between a categorical variable and a numerical or categorical variable.

```python
sns.barplot(data=df, x='day', y='tip')   #it shows an aggregated value, commonly the mean.
```

```python
sns.countplot(data=df,x='day')
```

```python
sns.boxplot(data=df,x='day',y='tip',hue='sex')
```

```python
sns.violinplot(data=df, x='day', y='tip')
```

```python
sns.stripplot(data=df,x='time',y='total_bill',hue='smoker')
```

```python
sns.swarmplot(data=df,x='time',y='total_bill',hue='smoker')
```

```python
sns.pointplot(data=df,x='day',y='tip')
```

**Distribution plots** are used to visualize how the values of a numerical variable are distributed, including their frequency, spread, center, and overall shape.

```python
sns.histplot(data=df,x='total_bill',hue='sex')
```

```python
sns.kdeplot(data=df, x='total_bill')
```

```python
sns.ecdfplot(data=df, x='total_bill')
```

```python
sns.rugplot(data=df, x='total_bill',height=0.6)
```

```python
sns.displot(data=df,x='total_bill',kind='hist')
```

Regression plots are used to visualize the relationship between variables and fit a regression line that shows the overall trend between them.

```python
sns.regplot(data=df,x='total_bill',y='tip')
```

```python
sns.lmplot(data=df,x='total_bill',y='tip')
```

```python
sns.residplot(data=df,x='total_bill',y='tip')
```

Matrix plots are used to visualize relationships or patterns among multiple numerical variables in a matrix-like format.

```python
sns.heatmap(df.corr(numeric_only=True),annot=True)
```

```python
sns.clustermap(df.corr(numeric_only=True), annot=True)
```

A multiplot grid is a layout that displays multiple plots together in rows and columns, making it easy to compare different groups or variables.

```python
g=sns.FacetGrid(df,col='day')
g.map_dataframe(sns.boxplot,x='sex',y='tip')
```

```python
g = sns.PairGrid(df[['total_bill', 'tip', 'size']])
g.map(sns.scatterplot)
```

```python
sns.pairplot(df[['total_bill', 'tip', 'size']])
```
