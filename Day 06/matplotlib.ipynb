```python
#univariant categorical
df["dept"]=["HR", "IT", "Sales", "Finance", "HR", "IT", "Sales", "Finance", "HR", "IT", "Sales", "Finance", "HR", "IT"]
```

```python
count=df["dept"].value_counts()
count
```

```python
plt.pie(count,labels=count.index,autopct="%1.1f",explode=[0,0.1,0,0])
plt.show()
```

```python
plt.bar(count.index,count,color=['red','green','blue','black'])
plt.show()
```

```python
#bivariate (numwrical - numerical)
data=pd.DataFrame({"Age": [22, 20, 26, 28, 54, 32, 34, 36, 38, 40, 42, 44, 46, 48],

"salary":[25000, 28000, 32000, 36000, 40000, 45000, 50000, 55000, 60000, 65000, 70000, 76000, 82000, 90000],
                   "Dept":["HR", "IT", "Sales", "Finance", "HR", "IT", "Sales", "Finance", "HR", "IT", "Sales", "Finance", "HR", "IT"]})
```

```python
data
```

```python
#scatter
plt.scatter(data['Age'],data['salary'],color="orange")
```

```python
#line
plt.plot(data['salary'],data['Age'],marker="*",linestyle=":",linewidth=2)
plt.grid()
plt.show()
```

```python
#bar chart
plt.bar(data["Age"],data['salary'],align="center")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")
plt.show()
```

```python
hr_sal=data[data['Dept']=="HR"]['salary']
hr_sal
```

```python
it_sal=data[data['Dept']=="IT"]['salary']
it_sal
```

```python
sales_Sal=data[data['Dept']=="Sales"]['salary']
sales_Sal
```

```python
plt.boxplot([it_sal,hr_sal,sales_Sal],tick_labels=['IT','HR','Sales'])

plt.show()
```

```python
salary_by_Dept=data.groupby('Dept')['salary'].sum()
salary_by_Dept
```

```python
plt.pie(salary_by_Dept,labels=salary_by_Dept.index,autopct="%1.1f")
plt.show()
```
